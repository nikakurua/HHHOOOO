```python
import pickle

class SimpleClass:
    """A simple example class to demonstrate pickling."""
    def __init__(self, value):
        self.value = value
    
    def display_value(self):
        print(f"The stored value is {self.value}.")

# Instantiate the class
obj = SimpleClass("Hello Pickle")

pickle_path = 'resources/simple_class.pkl'

# Pickle (serialize) the object
with open(pickle_path, 'wb') as f:
    pickle.dump(obj, f)

# Unpickle (deserialize) the object
with open(pickle_path, 'rb') as f:
    loaded_obj = pickle.load(f)

# Test the loaded object
loaded_obj.display_value()
```

# Introduction to NumPy

```python
import numpy as np
```

```python
arr_zeros = np.zeros((2, 3))         # 2x3 array of zeros
arr_ones = np.ones((3, 2), dtype=int)  # 3x2 array of ones with integer dtype
arr_random = np.random.random((3, 3)) # 3x3 array of random floats [0, 1)

print("Array of zeros:\n", arr_zeros, "\n")
print("Array of ones (int):\n", arr_ones, "\n")
print("Array of random floats:\n", arr_random)
```

```python
# Demonstrate indexing, slicing, shape, and summary functions.

arr = np.arange(1, 10).reshape((3, 3))  # 3x3 array with numbers 1 to 9
print("Original Array:\n", arr)
print("Shape of arr:", arr.shape)
print("Element at row 1, col 2:", arr[1, 2])  # 0-based indexing
print("First row:", arr[0, :])
print("Column 2:", arr[:, 2])

# Summaries
print("Sum of all elements:", arr.sum())
print("Mean of all elements:", arr.mean())
print("Standard Deviation:", arr.std())
print("Max element:", arr.max())
```

```python
# We'll use %%timeit, which works in a Jupyter cell, to compare 
# looping vs. NumPy vectorized multiplication.

# Large arrays for demonstration
large_arr_1 = np.random.rand(1000000)
large_arr_2 = np.random.rand(1000000)

# Method 1: Using Python for loop
def elementwise_multiply(a, b):
    result = np.empty_like(a)
    for i in range(len(a)):
        result[i] = a[i] * b[i]
    return result

# Method 2: Using NumPy vectorization
def numpy_multiply(a, b):
    return a * b

# Timeit comparisons:
# NOTE: The %%timeit magic command must be placed at the top of a cell 
# in an actual Jupyter Notebook.

# For illustration here, we show how you would do it:
```

```python
%%timeit -n 5 -r 5
_ = elementwise_multiply(large_arr_1, large_arr_2)
```

```python
%%timeit -n 5 -r 5
_ = numpy_multiply(large_arr_1, large_arr_2)
```

```python
# np.vectorize() can sometimes make code more readable, but 
# it doesn't always produce the same speedups as true vectorization.

def my_function(x):
    return x**2 + 2*x + 1

vec_my_function = np.vectorize(my_function)

data = np.random.rand(1000000)
```

```python
%%timeit -n 5 -r 5
result_vec = vec_my_function(data)
```

```python
%%timeit -n 5 -r 5
_ = my_function(data)
```

```python
data = data.tolist()
```

```python
%%timeit -n 5 -r 5
for x in data:
    x = x ** 2 + 2 * x + 1
```

```python
# A common benchmark is multiplying large matrices. 
# Here we'll create three 4096x4096 matrices and multiply them in sequence:
# Result = A x B x C. 
# We'll demonstrate standard NumPy matrix multiplication using '@' operator.

matrix_size = 4096

# Generating random matrices
A = np.random.rand(matrix_size, matrix_size)
B = np.random.rand(matrix_size, matrix_size)
C = np.random.rand(matrix_size, matrix_size)
```

```python
%%timeit -n 1 -r 1

result = A @ B @ C
```

```python
arr = np.arange(1, 21)  # 1D array from 1 to 20
mask = arr % 2 == 0     # True where elements are even, False otherwise

print("Original array:", arr)
print("Mask (True for even numbers):", mask)

# Masking usage
even_numbers = arr[mask]
print("Even numbers:", even_numbers)

# Slicing usage
print("Elements from index 2 to 5:", arr[2:6])   # slice up to but not including 6
print("Last 5 elements:", arr[-5:])
print("Every 2nd element:", arr[::2])
```

```python
# Usage: np.where(condition, value_if_true, value_if_false)

arr = np.array([10, 15, 20, 25, 30])
print("Original array:", arr)

# Replace values >= 20 with 999, otherwise keep original
modified = np.where(arr >= 20, 999, arr)
print("Modified array (values >= 20 -> 999):", modified)

# We can also use it to find indices where a condition is True.
indices = np.where(arr >= 20)
print("Indices where arr >= 20:", indices)
```

# Pandas

```python
# Load the Titanic dataset and explore basic DataFrame functionality.

titanic_df = pd.read_csv('resources/titanic/train.csv')  # or provide the correct path
print("Loaded Titanic dataset with shape:", titanic_df.shape)

# Convert a single column to a Series
age_series = titanic_df['Age']
print("\nPandas Series (Age) example:\n", age_series.head())

# Display DataFrame basics
display(titanic_df.head())
print("\nInfo:")
titanic_df.info()
print("\nDescribe:")
display(titanic_df.describe())
```

```python
older_passengers = titanic_df[titanic_df['Age'] > 50]
print("Number of passengers older than 50:", len(older_passengers))

# Sort by Fare descending
sorted_by_fare = titanic_df.sort_values(by='Fare', ascending=False)
print("\nTop 5 rows sorted by Fare descending:")
display(sorted_by_fare.head())

# Select a subset of columns
subset = titanic_df[['Name', 'Age', 'Survived']]
print("\nSubset of the DataFrame:")
display(subset.head())
```

```python
# We'll demonstrate iterrows(), itertuples(), apply(), and vectorized operations 
# by performing a simple operation (e.g., add a new 'AgePlusOne' column).

# 1) iterrows(): returns index, Series
def add_one_iterrows(df):
    result = []
    for idx, row in df.iterrows():
        result.append(row['Age'] + 1 if pd.notnull(row['Age']) else None)
    return result

# 2) itertuples(): returns named tuples
def add_one_itertuples(df):
    result = []
    for row in df.itertuples():
        result.append(row.Age + 1 if pd.notnull(row.Age) else None)
    return result

# 3) apply(): apply function along axis of DataFrame
def add_one_apply(age):
    return age + 1 if pd.notnull(age) else None

# 4) Vectorized operation:
#    We can directly add 1 to the 'Age' column, using fillna() to handle NaNs.

# We'll use %timeit to compare them:
```

```python
%%timeit -n 1 -r 1
add_one_iterrows(titanic_df)
```

```python
%%timeit -n 1 -r 1
add_one_itertuples(titanic_df)
```

```python
%%timeit -n 1 -r 1
titanic_df['Age'].apply(add_one_apply)
```

```python
%%timeit -n 1 -r 1
titanic_df['Age'] + 1
```

```python
# Example: create a new datetime column, manipulate time, etc.
# For Titanic, we don't have a date column by default, 
# but let's show how we might parse a hypothetical 'Date' column if it existed.

# We'll create a fake date column just for demonstration
import numpy as np
import datetime

# Create a date range
num_rows = len(titanic_df)
random_days = np.random.randint(1, 30, size=num_rows)
start_date = datetime.datetime(1912, 4, 1)

titanic_df['FakeDate'] = [start_date + datetime.timedelta(days=int(x)) for x in random_days]

# Convert to datetime if it wasn't already
titanic_df['FakeDate'] = pd.to_datetime(titanic_df['FakeDate'])

# Extract year, month, day
titanic_df['Year'] = titanic_df['FakeDate'].dt.year
titanic_df['Month'] = titanic_df['FakeDate'].dt.month
titanic_df['Day'] = titanic_df['FakeDate'].dt.day

print(titanic_df[['FakeDate', 'Year', 'Month', 'Day']].head())
```

```python
from tqdm.notebook import tqdm
tqdm.pandas()

titanic_df.Survived.progress_apply(lambda x: x > 0)
```

```python
# Simple built-in plotting with DataFrame.plot() 
# We'll plot the distribution of Ages for survivors vs. non-survivors.

import matplotlib.pyplot as plt

# Drop missing ages
df_age = titanic_df.dropna(subset=['Age'])

plt.figure(figsize=(8, 5))
df_age[df_age['Survived'] == 1]['Age'].plot(kind='hist', bins=20, alpha=0.6, label='Survived', color='blue')
df_age[df_age['Survived'] == 0]['Age'].plot(kind='hist', bins=20, alpha=0.6, label='Did Not Survive', color='red')
plt.title('Age Distribution by Survival Status')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

```python
# -----------------------------------------------------
# Plot the average Fare over different passenger classes as a simple line plot.

titanic_line_data = titanic_df.groupby('Pclass')['Fare'].mean()
titanic_line_data.plot(kind='line', title='Average Fare by Passenger Class (Line Plot)', marker='o')
plt.ylabel('Average Fare')
plt.xlabel('Passenger Class')
plt.show()
```

```python
# -----------------------------------------------------
# Plot a bar chart showing number of passengers by Embarked port.

titanic_bar_data = titanic_df['Embarked'].value_counts()
titanic_bar_data.plot(kind='bar', title='Number of Passengers by Embarked Port', color='skyblue')
plt.ylabel('Count')
plt.xlabel('Port')
plt.show()
```

```python
# Show survival count by Sex in a horizontal bar plot.

titanic_hbar_data = titanic_df.groupby('Sex')['Survived'].sum()  # total survivors by sex
titanic_hbar_data.plot(kind='barh', title='Total Survivors by Sex', color='orange')
plt.xlabel('Number of Survivors')
plt.ylabel('Sex')
plt.show()
```

```python
# We can use an area plot on a grouped dataset to show cumulative trends.
# For demonstration, let's group by Pclass and sum up 'Fare', 'SibSp', 'Parch'.

titanic_area_data = titanic_df.groupby('Pclass')[['Fare', 'SibSp', 'Parch']].sum()
titanic_area_data.plot(kind='area', title='Sum of Fare, SibSp, and Parch by Passenger Class', alpha=0.7)
plt.xlabel('Passenger Class')
plt.ylabel('Sum of Values')
plt.show()
```

```python
# We can do a scatter plot of Age vs. Fare using the DataFrame's built-in plot functionality.

titanic_df.plot(kind='scatter', x='Age', y='Fare', alpha=0.5, color='green', title='Scatter Plot of Age vs. Fare')
plt.show()
```

```python
# -----------------------------------------------------
# We can also directly call box plot on a subset of numeric columns.

numeric_cols = ['Age', 'Fare', 'SibSp', 'Parch']
titanic_df[numeric_cols].plot(kind='box', title='Box Plots of Age, Fare, SibSp, and Parch')
plt.show()
```

```python
# Plot pie chart for class distribution.

pclass_counts = titanic_df['Pclass'].value_counts()
pclass_counts.plot(kind='pie', autopct='%1.1f%%', title='Passenger Class Distribution', startangle=140, figsize=(5,5))
plt.ylabel('')  # Hide the default y-label
plt.show()
```
